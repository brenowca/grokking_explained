```python
import pandas as pd
import numpy as np
from os.path import join as path_join
```

```python
def downsample_dataframe(df, num_samples, column='class'):
        # Check if the 'class' column exists in the dataframe
    if column not in df.columns:
        raise ValueError(f"The dataframe does not contain a '{column}' column.")

    # Get unique classes
    classes = df[column].unique()

    # Initialize an empty list to hold the down-sampled dataframes
    downsampled_dfs = []

    # Loop through each class and down-sample
    for cls in classes:
        df_class = df[df[column] == cls]
        
        # Ensure there are enough samples in the class to down-sample
        if len(df_class) < num_samples:
            raise ValueError(f"Not enough samples in class {cls} to down-sample to {num_samples} samples.")
        
        # Randomly select num_samples indices from the class
        sampled_indices = np.random.choice(df_class.index, num_samples, replace=False)
        downsampled_df_class = df_class.loc[sampled_indices]
        
        downsampled_dfs.append(downsampled_df_class)

    # Concatenate all down-sampled dataframes
    downsampled_df = pd.concat(downsampled_dfs)

    return downsampled_df
```

```python
root_folder = "/dccstor/kcsys/brenow/repos/grokking/datasets"
dest_folder = "/dccstor/kcsys/brenow/repos/grokking/datasets_down"
```

## Equivariant dataset

```python
full_dataset_name = "V0_binaryTreeMultiDim_distSuperClasses_ns_10000_ndg_9_ndpg_13.csv"
test_dataset_name = "V0_binaryTreeMultiDim_distSuperClasses_ns_10000_ndg_9_ndpg_13_test.csv"
train_dataset_name = "V0_binaryTreeMultiDim_distSuperClasses_ns_10000_ndg_9_ndpg_13_train.csv"

num_samples = 20000//40
column='class'
```

```python
print("Reading test dataset")
test_dataset = pd.read_csv(path_join(root_folder, test_dataset_name))
print(f"Test dataset: {len(test_dataset)} and {test_dataset.columns}")
print(f"Test class: {sorted(test_dataset['class'].unique())} and \n\t super_class{sorted(test_dataset['super_class'].unique())}")
```

```python
print("Reading train dataset")
train_dataset = pd.read_csv(path_join(root_folder, train_dataset_name))
print(f"Train dataset: {len(train_dataset)} and {train_dataset.columns}")
print(f"Train class: {sorted(train_dataset['class'].unique())} and \n\t super_class{sorted(train_dataset['super_class'].unique())}")
```

```python
print("Reading full dataset")
full_dataset = pd.read_csv(path_join(root_folder, full_dataset_name))
print(f"Full dataset: {len(full_dataset)} and {full_dataset.columns}")
print(f"Full class: {sorted(full_dataset['class'].unique())} and \n\t super_class{sorted(full_dataset['super_class'].unique())}")
```

```python
print("Downsampling test dataset")
downsampled_test_dataset = downsample_dataframe(test_dataset, num_samples, column=column)

print(f"Writing test dataset to {dest_folder}")

samples_per_class = downsampled_test_dataset.groupby([column])[column].count()
print(f"Samples per class: {samples_per_class.iloc[0]}")
assert np.all(samples_per_class == samples_per_class.iloc[0]), "Sampling got wrong number of samples"

downsampled_test_dataset.to_csv(path_join(dest_folder, test_dataset_name))
```

```python
print("Downsampling train dataset")
downsampled_train_dataset = downsample_dataframe(train_dataset, num_samples, column=column)

samples_per_class = downsampled_train_dataset.groupby([column])[column].count()
print(f"Samples per class: {samples_per_class.iloc[0]}")
assert np.all(samples_per_class == samples_per_class.iloc[0]), "Sampling got wrong number of samples"

print(f"Writing train dataset to {dest_folder}")
downsampled_train_dataset.to_csv(path_join(dest_folder, train_dataset_name))
```

```python
print("Downsampling full dataset")
downsampled_full_dataset = downsample_dataframe(full_dataset, num_samples, column=column)

samples_per_class = downsampled_full_dataset.groupby([column])[column].count()
print(f"Samples per class: {samples_per_class.iloc[0]}")
assert np.all(samples_per_class == samples_per_class.iloc[0]), "Sampling got wrong number of samples"

print(f"Writing full dataset to {dest_folder}")
downsampled_full_dataset.to_csv(path_join(dest_folder, full_dataset_name))
```
